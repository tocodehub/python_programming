```python
import xml.etree.ElementTree as et
tree = et.ElementTree(file='menu.xml')
root = tree.getroot()
```

```python
print(root.tag)
```

```python
type(root)
```

```python
for child in root:
    print('tag:', child.tag, 'attributes:', child.attrib)
    for grandchild in child:
        print('\ttag:', grandchild.tag, 'attributes:', grandchild.attrib)
```

```python
len(root)        # 3
```

```python
len(root[0])     # 2
```

```python
type(root[0])
```

```python
print(root[0])              # root의 첫번째 자식 : Element 'breakfast'
print(root[0].get("hours")) # get() 속성 값 : 'breakfast'의 'hours' 속성의 값 리턴
print(root[0].keys())       # 모든 속성의 키를 리스트로 리턴
print(root[0].items())      # 모든 속성의 key, value 쌍을 리턴
```

```python
lunch = root.find("lunch")
print(lunch)
print(lunch.get("hours"))
print(lunch.keys())
print(lunch.items())
```

```python
print(root[0].findall('item'))
```

```python

for item in root[0].findall('item'):
    print(item.get('price'))
    print(item.text)
```

```python
import json
j1 = {"name":"홍길동", "birth":"0525", "age": 30}
j1
# {'age': 30, 'birth': '0525', 'name': '홍길동'}
json.dumps(j1)
# '{"age": 30, "birth": "0525", "name": "\\ubc15\\uc751\\uc6a9"}'
```

```python
print(json.dumps(j1, indent=2))
# {
#   "age": 30,
#   "birth": "0525",
#   "name": "\ubc15\uc751\uc6a9"
# }
```

```python

json.dumps([1,2,3])    # '[1, 2, 3]'
json.dumps((4,5,6))    # '[4, 5, 6]'
```

```python
j1 = {"name":"홍길동", "birth":"0525", "age": 30}
d1 = json.dumps(j1)
json.loads(d1)
# {'name': '홍길동', 'birth': '0525', 'age': 30}
```

```python
with open('./myinfo.json') as f:
    data = json.load(f)

print(type(data))    # <class 'dict'>
print(data)          # {'name': '홍길동', 'birth': '0525', 'age': 30}
```
